## 1. Strings

A **string** is a sequence of characters enclosed in single `'...'` or double `"..."` quotes. Strings are **immutable** — once created, they cannot be changed in place.

### Key concepts
- **Indexing**: Access individual characters using `s[index]` (0-based, negatives count from the end)
- **Slicing**: Extract substrings using `s[start:stop:step]`
- **Methods**: Built-in functions like `.upper()`, `.lower()`, `.strip()`, `.split()`, `.replace()`, `.count()`, `.find()`
- **`.format()`**: Classic string interpolation using `"Hello {}".format(name)` or named placeholders `"{name}".format(name=name)`
- **f-strings**: Modern, readable string interpolation using `f"Hello {name}"` (preferred in Python 3.6+)

````python
# ── Strings: Demo ──────────────────────────────────────────────

s = "Hello, Python!"

# Indexing
print("First char :", s[0])        # H
print("Last char  :", s[-1])       # !

# Slicing  [start : stop : step]
print("Slice [0:5]:", s[0:5])      # Hello
print("Reversed   :", s[::-1])     # !nohtyP ,olleH

# Common methods
print("Upper      :", s.upper())
print("Replace    :", s.replace("Python", "World"))
print("Split      :", s.split(", "))

# f-strings
name = "Alice"
age  = 30
print(f"Name: {name}, Age: {age}, Next year: {age + 1}")

# .format() — classic string interpolation
greeting = "Hello, {}! You are {} years old.".format(name, age)
print("format()   :", greeting)

# Named placeholders
template = "Name: {n}, Age: {a}".format(n=name, a=age)
print("named fmt  :", template)

# .count() — occurrences of a substring
sentence = "the cat sat on the mat"
print("count('at'):", sentence.count("at"))   # 3
print("count('the'):", sentence.count("the")) # 2

# .find() — index of first occurrence (-1 if not found)
print("find('cat') :", sentence.find("cat"))  # 4
print("find('dog') :", sentence.find("dog"))  # -1  (not found)
````

## 2. Integers & Floats

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42`, `-7`, `0` | Whole numbers, unlimited precision |
| `float` | `3.14`, `-0.5`, `2.0` | Decimal numbers (IEEE 754 double precision) |

### Key operations
| Operator | Meaning | Example |
|----------|---------|--------|
| `+` `-` `*` | Add, subtract, multiply | `3 + 4 → 7` |
| `/` | True division (always float) | `7 / 2 → 3.5` |
| `//` | Floor division | `7 // 2 → 3` |
| `%` | Modulo (remainder) | `7 % 2 → 1` |
| `**` | Exponentiation | `2 ** 8 → 256` |

````python
# ── Integers & Floats: Demo ────────────────────────────────────

a, b = 17, 5

print("Addition      :", a + b)    # 22
print("True division :", a / b)    # 3.4
print("Floor division:", a // b)   # 3
print("Modulo        :", a % b)    # 2
print("Exponent      :", 2 ** 10)  # 1024

# Type conversion
pi = 3.14159
print("float → int   :", int(pi))         # 3  (truncates, not rounds)
print("int → float   :", float(42))       # 42.0
print("str → int     :", int("100") + 1)  # 101

# Useful built-ins
print("abs(-9)       :", abs(-9))
print("round(3.567,2):", round(3.567, 2))
print("type(pi)      :", type(pi))
````

## 3. Lists

A **list** is an **ordered, mutable** collection that can hold items of any type.

```python
fruits = ["apple", "banana", "cherry"]
```

| Operation | Code | Result |
|-----------|------|--------|
| Access | `fruits[1]` | `'banana'` |
| Append | `fruits.append('date')` | adds to end |
| Insert | `fruits.insert(1, 'avocado')` | inserts at index |
| Remove | `fruits.remove('banana')` | removes by value |
| Pop | `fruits.pop()` | removes & returns last |
| Sort | `fruits.sort()` | sorts in-place |
| Length | `len(fruits)` | count of items |

````python
# ── Lists: Demo ────────────────────────────────────────────────

nums = [3, 1, 4, 1, 5, 9, 2, 6]

print("Original  :", nums)
print("Length    :", len(nums))
print("Max / Min :", max(nums), "/", min(nums))
print("Sorted    :", sorted(nums))     # new list, original untouched

nums.append(7)
print("After append(7):", nums)

nums.remove(1)                         # removes FIRST occurrence
print("After remove(1):", nums)

# Slicing works just like strings
print("First three    :", nums[:3])
print("Every other    :", nums[::2])

# List comprehension — concise way to build lists
squares = [x ** 2 for x in range(1, 6)]
print("Squares 1-5    :", squares)
````

## 4. Tuples

A **tuple** is an **ordered, immutable** collection — like a list that cannot be changed after creation. Use tuples for data that should stay constant (e.g. coordinates, RGB values, database rows).

```python
point = (10, 20)        # parentheses (optional but conventional)
single = (42,)          # trailing comma required for single-item tuple!
```

- Supports **indexing** and **slicing** (same as lists)
- Supports **unpacking**: `x, y = point`
- Slightly **faster** than lists for iteration

````python
# ── Tuples: Demo ───────────────────────────────────────────────

rgb_red = (255, 0, 0)
print("Tuple    :", rgb_red)
print("Red value:", rgb_red[0])

# Unpacking
r, g, b = rgb_red
print(f"R={r}, G={g}, B={b}")

# Tuples are immutable — this would raise a TypeError:
# rgb_red[0] = 100  ← TypeError!

# Useful: swap variables without a temp variable
x, y = 5, 10
x, y = y, x
print(f"After swap: x={x}, y={y}")

# Named tuple — adds clarity
from collections import namedtuple
Point = namedtuple("Point", ["x", "y"])
p = Point(3, 7)
print(f"Point: {p}, x={p.x}, y={p.y}")
````

## 5. Sets

A **set** is an **unordered collection of unique elements**. Great for membership testing and eliminating duplicates.

```python
s = {1, 2, 3, 2, 1}   # → {1, 2, 3}  duplicates removed automatically
empty = set()           # NOT {} — that's an empty dict!
```

| Operation | Code | Meaning |
|-----------|------|---------|
| Union | `A \| B` | all elements in A or B |
| Intersection | `A & B` | elements in both A and B |
| Difference | `A - B` | in A but not in B |
| Symmetric diff | `A ^ B` | in A or B but not both |

````python
# ── Sets: Demo ─────────────────────────────────────────────────

# Deduplication
raw = [1, 2, 2, 3, 3, 3, 4]
unique = set(raw)
print("Unique items  :", unique)

# Set operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Union     A|B :", A | B)
print("Intersect A&B :", A & B)
print("Diff      A-B :", A - B)
print("Sym. diff A^B :", A ^ B)

# Fast membership test  O(1) vs O(n) for lists
print("3 in A        :", 3 in A)
print("9 in A        :", 9 in A)

# Mutating a set
A.add(10)
A.discard(1)   # discard is safe — no error if element missing
print("After mutations:", A)
````
